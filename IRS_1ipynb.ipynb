# similarity


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
doc1 = "Machine learning is  a branch of artificial intelligence that focuses on building aapplications that learn from data."
doc2 ="Artifical intellegence and machine learning are widely used to make smart applications based on data."
doc3 ="in future Artifical intelligence and machine learning domains are not going to work because of changing the generation"
documents = [doc1, doc2, doc3]
Vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = Vectorizer.fit_transform(documents)
simaliraty_score = cosine_similarity(tfidf_matrix[0:1],tfidf_matrix[1:2])[0][0]
print("Cosine Similarity between Document 1 and Document 2 and document 3:")
print(f"{simaliraty_score:.4f}")
```
